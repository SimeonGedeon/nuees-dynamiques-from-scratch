```python
# Cellule 1 — imports
import sys, os
sys.path.insert(0, "..")
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons
from src import NueesDynamiques, NoyauPoint, NoyauMultiPoints, NoyauAxe

# Cellule 2 — données
X, _ = make_blobs(n_samples=500, centers=4, cluster_std=0.8, random_state=0)

# Cellule 3 — entraînement
model = NueesDynamiques(k=4, noyau_factory=lambda: NoyauPoint(), seed=0, verbose=True)
model.fit(X)

# Cellule 4 — convergence
plt.plot(model.inertia_history, marker="o")
plt.xlabel("Itération"); plt.ylabel("W")
plt.title("Convergence"); plt.grid(alpha=0.3); plt.show()

# Cellule 5 — clusters
plt.scatter(X[:,0], X[:,1], c=model.labels, cmap="tab10", s=12)
plt.title("Clusters (noyau point)"); plt.show()

# Cellule 6 — comparaison des noyaux
# (idem script run_demo.py)
```
